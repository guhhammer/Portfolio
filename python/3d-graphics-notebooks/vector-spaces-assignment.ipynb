### 1) History of vector spaces

- ***Vector spaces began with the complex numbers, which were discovered by accident. After the discovery, the mathematician Hamilton tried to group them in 3 dimensions, but that was not enough: he needed 4, and during a walk on a bridge near his home he had an idea and invented the quaternions.***<br><br>

     ***The story was so significant that today the bridge where Hamilton invented the quaternions is called Quaternion Bridge.***<br><br>

     ***The pragmatic Gibbs and Heaviside took the scalar and vector products and created the three-dimensional system we use to this day.***<br><br>

     ***Later, Grassmann developed a rather simple geometric product. With Clifford's improvement of that product, it would allow the generalisation of the complex numbers, of the quaternions, of the products of vector calculus and of anything else to any dimension.***

     ***We can conclude that if the complex numbers and the quaternions precede the Gibbs-Heaviside vector system, the system of Grassmann and Clifford is its successor, and the more time passes the more implementations will come, and we should use them even more.***<br><br>

     ***Cayley went far beyond everyone: inspired by the old ideas about vectors, he invented the octonions, whose product was not limited to the complex numbers and quaternions. Even as a matrix product, it cannot directly reproduce the product of the octonions.***<br><br>

     ***Among many later mathematicians, the vectors studied to this day were created, making history up to the present.*** <br><br>

### 2) Definition of vector space

##### Concept:
  - The common notion of vectors as objects with magnitude, direction and orientation, together with the operations of addition and multiplication by real numbers, forms the basic idea of a vector space.
  - A field $K$, i.e. a set with two internal operations with distributive properties, whose elements are called scalars. The real numbers, with addition and multiplication, are an example of a field.

##### Operations:
- Let V be a non-empty set on which addition and multiplication by a scalar are defined, that is:<br><br>
- Addition:<br>
$\forall u,  v \in V, u + v \in V   $<br>

- Multiplication:<br>
$\forall \alpha \in \Re , \forall u \in V,  \alpha u \in V$

#### Properties of the operations:<br>
- For addition:<br>
a1:$ u + (v + w) = (u + v) + w$  <br>
a2:$ u + v = v + u $<br>
a3:$ u + 0 = u $<br>
a4:$ u + (-u) = 0 $<br><br>
- For multiplication: <br>
m1: $\alpha (\beta*u) = (\alpha *\beta)*u$ <br>
m2: $(\alpha + \beta)*u = \alpha*u + \beta*u $<br>
m3: $\alpha*(u+v) = \alpha*u + \alpha*v $<br>
m4: $1 * (u) = u$ <br>

### 3) Examples of vector spaces

###### **Vector spaces have subdivisions that classify their types and applications:**

- A Euclidean vector space is a real space with a finite number of dimensions and an inner product (a function of two vectors satisfying certain axioms).
- A Hilbert space is defined when a space has a metric generated by the inner product, making it a complete space.
- A normed space is a vector space with a defined norm, i.e. a function that assigns a non-negative real number to each vector.
- A Banach space is a normed vector space in which the real and the complex numbers form the same, the norm being their absolute value.
- A topological vector space combines the concepts of convergence, continuity and connectedness, so that the operations are defined continuously on the vector space.  

##### Note: 

  - 
The elements of a vector space V are called vectors, whether they are polynomials, matrices or numbers. These elements have different natures but behave the same way under addition and scalar multiplication, as if we were working with the vectors of $\Re ^2$ and $\Re ^3$ themselves.

### 4) Applications of vector spaces.

- The spectral colour space is a vector space of dimension 3 (the three primary colours):
  - Different coordinate systems (known as colour systems) are considered in this space, according to the application or the graphics output device (monitor, printer, etc.).
  - It is often necessary to move from one coordinate system to another, which is done with a change-of-coordinates matrix. For example, the matrix from the RGB system to the XYZ system is a 3x3 matrix obtained by taking white as a fixed point of the transformation.
<br><br>
- An example of using vectors would be the calculations of an aircraft, specifically of a turbine, placing pressure vectors for the air intake and the exhaust, which is very important for the organisation and harmony of the whole assembly.
<br><br>
- They can also be used to compute the pressure on the aircraft at a given altitude, so that no accidents or depressurisation happen and nothing happens to the passengers
