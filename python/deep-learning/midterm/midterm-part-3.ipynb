```
"""
    A.	Read the heart disease data from URL:
    https://raw.githubusercontent.com/Marcelpv96/Heart-Disease-UCI/master/data/heart.csv
"""
import pandas as pd

data = pd.read_csv("https://raw.githubusercontent.com/Marcelpv96/Heart-Disease-UCI/master/data/heart.csv")

"""
    A.	Please use all the independent variables and create a Model for predicting the heart disease (target).
"""

import numpy as np

print("Data header size: ",len(list(data)), "\nData header: ",list(data))
input_ = np.array([ [data.iloc[:, 0][i],data.iloc[:, 1][i],data.iloc[:, 2][i],data.iloc[:, 3][i],data.iloc[:, 4][i],data.iloc[:, 5][i],
                    data.iloc[:, 6][i],data.iloc[:, 7][i],data.iloc[:, 8][i],data.iloc[:, 9][i],data.iloc[:, 10][i],data.iloc[:, 11][i],
                    data.iloc[:, 12][i]] for i in range(0, len(data.iloc[:, 0]))], dtype=float)

output_ = np.array( [ data.iloc[:, 13][i] for i in range(0, len(data.iloc[:, 13])) ], dtype=float)


from sklearn.tree import DecisionTreeClassifier

# Making a Tree-Decision Model:
model = DecisionTreeClassifier()

model = model.fit(input_, output_)

```

```
"""
    B.	Using your model predict disease for the following two persons. 
    (The input features are given, you need to fill the target):
"""

person1 = np.array([37, 0, 0, 129, 245,	0, 1,	183, 1,	0.6, 1,	0, 2], dtype=float).reshape(1,-1)	
person2 = np.array([54,	1, 2,	145, 354,	0, 1,	153, 0,	1.4, 1,	0, 1], dtype=float).reshape(1,-1)

from tabulate import tabulate

headers = ['age','sex','cp','trestbps','chol','fbs','restecg','thalach','exang','oldpeak','slope','ca','thal','target']

person1_t = [37, 0, 0, 129, 245,	0, 1,	183, 1,	0.6, 1,	0, 2, int(model.predict(person1)[0])]
person2_t = [54,	1, 2,	145, 354,	0, 1,	153, 0,	1.4, 1,	0, 1, int(model.predict(person2)[0])]

print(tabulate([person1_t, person2_t], headers=headers))
```
