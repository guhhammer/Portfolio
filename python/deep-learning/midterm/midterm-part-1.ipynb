```
""" 
   A. Your Word frequency counter shall read a text file from URL
   https://www.gutenberg.org/files/2701/2701.txt as an input.
"""

import urllib

input = urllib.request.urlopen("https://www.gutenberg.org/files/2701/2701.txt")

lines = [line.decode("utf-8")[:-2] for line in input]
```

```
"""
    B. Your Word Frequency counter shall display the following information:
      - Number of characters in the file:
      - Number of words in the file:
      - Number of sentences in the file:
      - Number of lines in the file:
"""

n_characters = sum([len(line) for line in lines]) # Number of characters.
n_words = sum([len(line.split()) for line in lines]) # Number of words.


import re   # Number of sentences (I'm using regex to break the text into sentences).
n_sentences = len(re.split(r' *[\.\?!][\'"\)\]]* *', " ".join(lines))) 


n_lines = len(lines)  # Number of lines.

import warnings
warnings.filterwarnings("ignore")

print(("\t- Number of characters in the file: %d"
       "\n\t- Number of words in the file: %d"
       "\n\t- Number of sentences in the file: %d"
       "\n\t- Number of lines in the file: %d") 
       % (n_characters, n_words, n_sentences, n_lines))

```

```
"""
    C.	Your Word Frequency counter shall count the frequency (number of appearances)
    of each word appeared in the input file.  Sort the result from the most used word
    to the least used word and display the first 10 words (the 10 most used words).
    The display format shall be as the following (each line includes a word and its
    frequency, separated by a comma):
            
              Word1, frequency, 
              Word2, frequency 
              Word3, frequency 
"""

def word_frequency_counter(number_of_words_to_display):
    
    frequency_counter = dict() # Make a dictionary.
    
    words = [line.split() for line in lines] # Get Words from the lines array(defined in cell A).

    # Fill the dictionary:
    for string in range(0, len(words)):
        for word in range(0, len(words[string])):
            if words[string][word] in frequency_counter:
                frequency_counter[ words[string][word] ] += 1
            else:
                frequency_counter[ words[string][word] ] = 1


    import numpy as np # Sort the dictionary values to get the greater values.
    arg_values = np.sort(np.array([i for i in frequency_counter.values()]))[::-1]

    # Get the number of values to display. 
    highest_value_occurancies = [arg_values[i] for i in range(0, number_of_words_to_display)]

    answers = [ [word, frequence]
                for i in range(0, len(highest_value_occurancies))
                for word, frequence in frequency_counter.items()
                 if frequence == highest_value_occurancies[i] ]

    import warnings
    warnings.filterwarnings("ignore")

    print("\n\tDisplaying the top %d most frequent words in the file:\n" % number_of_words_to_display)
    for answer in answers:
        print("\t\t%s, %d" % (answer[0], answer[1]))
    print("\n\n")


word_frequency_counter(10)
```
