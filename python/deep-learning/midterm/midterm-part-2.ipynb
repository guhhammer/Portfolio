```
"""
    A.	Read the World Happiness Data (2017)  from URL 
    https://raw.githubusercontent.com/ebubekiraygun/World-Happiness-Data-on-Python/master/2017.csv
"""
import pandas as pd

data = pd.read_csv("https://raw.githubusercontent.com/ebubekiraygun/World-Happiness-Data-on-Python/master/2017.csv")
```

```
"""
    B.	Please write a code which generates the plot - 
    GDP.per.Capita vs. Happiness.Score  (similar to the following)
"""
print("Data Header: ",list(data),end="\n\n")

gdp_per_capita_index = [i for i in range(0, len(list(data))) if list(data)[i].lower() == "economy..gdp.per.capita."][0]
happiness_score_index = [i for i in range(0, len(list(data))) if list(data)[i].lower() == "happiness.score"][0] 

print("GDP per Capita index: %d\nHappiness Score index: %d\n" %(gdp_per_capita_index, happiness_score_index))

import numpy as np

gdp_per_capita_array = np.array(data.iloc[:, gdp_per_capita_index], dtype=float)
happiness_score_array = np.array(data.iloc[:, happiness_score_index], dtype=float)


import matplotlib.pyplot as plt

plt.title("Gustavo Hammerschmidt - GDP.per.Capita vs. Happiness.Score\n")
plt.plot(gdp_per_capita_array, happiness_score_array, "o", label='GDP.per.Capita')
plt.legend(loc='upper left')
plt.xlabel("GDP per Capita")
plt.ylabel("Happiness Score")
plt.show()

```

```
"""
    C.	Please write a code which generates the plot - Freedom vs. Happiness.Score  (similar to the following)
"""

print("Data Header: ",list(data),end="\n\n")

freedom_index = [i for i in range(0, len(list(data))) if list(data)[i].lower() == "freedom"][0]
happiness_score_index = [i for i in range(0, len(list(data))) if list(data)[i].lower() == "happiness.score"][0] 

print("Freedom index: %d\nHappiness Score index: %d\n" %(gdp_per_capita_index, happiness_score_index))

import numpy as np

freedom_array = np.array(data.iloc[:, freedom_index], dtype=float)
happiness_score_array = np.array(data.iloc[:, happiness_score_index], dtype=float)


import matplotlib.pyplot as plt

plt.title("Gustavo Hammerschmidt - Freedom vs. Happiness.Score\n")
plt.plot(freedom_array, happiness_score_array, "o", color="green", label='Freedom')
plt.legend(loc='upper left')
plt.xlabel("Freedom")
plt.ylabel("Happiness Score")
plt.show()

```
